## 1. Defining Basic Functions

Functions are reusable blocks of code defined using the `def` keyword. They help organize code and avoid repetition.

```python
# Basic function definition
def greet():
    print("Hello, World!")

greet()  # Calling the function

# Function with a parameter
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
```

## 2. Function Parameters and Arguments

- **Parameters**: Variables listed in the function definition
- **Arguments**: Values passed when calling the function
- **Positional arguments**: Matched by position
- **Keyword arguments**: Matched by name

```python
# Positional arguments - order matters
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}")

describe_pet("dog", "Max")  # Positional

# Keyword arguments - order doesn't matter
describe_pet(name="Whiskers", animal="cat")

# Mixed: positional first, then keyword
describe_pet("hamster", name="Fluffy")
```

## 3. Default Parameter Values

Default values are assigned when the function is defined. Parameters with defaults must come after parameters without defaults.

```python
# Default parameter values
def power(base, exponent=2):
    return base ** exponent

print(power(5))       # Uses default exponent=2 → 25
print(power(5, 3))    # Overrides default → 125

# Multiple defaults
def create_user(name, role="user", active=True):
    return {"name": name, "role": role, "active": active}

print(create_user("Alice"))
print(create_user("Bob", role="admin"))
```

## 4. Variable-Length Arguments (*args and **kwargs)

- `*args`: Collects extra positional arguments as a tuple
- `**kwargs`: Collects extra keyword arguments as a dictionary

```python
# *args - variable positional arguments
def sum_all(*args):
    print(f"Args received: {args}")
    return sum(args)

print(sum_all(1, 2, 3, 4, 5))

# **kwargs - variable keyword arguments
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=30, city="Toronto")

# Combining both
def flexible_func(required, *args, **kwargs):
    print(f"Required: {required}")
    print(f"Args: {args}")
    print(f"Kwargs: {kwargs}")

flexible_func("hello", 1, 2, 3, x=10, y=20)
```

### Understanding **kwargs in Detail

**`**kwargs`** stands for **"keyword arguments"**. It allows a function to accept any number of keyword arguments (arguments passed by name).

**How it works:**
- When you call a function with `key=value` pairs, `**kwargs` collects them into a **dictionary**
- The `**` operator unpacks the dictionary when calling another function
- The name `kwargs` is just a convention—you can use any name (e.g., `**options`)

**Key Points:**
| Feature | Description |
|---------|-------------|
| Type | Dictionary (`dict`) |
| Access | Use `kwargs['key']` or `kwargs.get('key')` |
| Iteration | Loop with `.items()`, `.keys()`, `.values()` |
| Order | Must come AFTER `*args` in function definition |

```python
# Example 1: Basic **kwargs usage
def greet_user(**kwargs):
    print(f"kwargs is a dictionary: {type(kwargs)}")
    print(f"Contents: {kwargs}")
    
    # Access values safely with .get()
    name = kwargs.get('name', 'Guest')
    greeting = kwargs.get('greeting', 'Hello')
    print(f"{greeting}, {name}!")

greet_user(name="Alice", greeting="Welcome")
print()
greet_user()  # Uses defaults
```

```python
# Example 2: Iterating through kwargs
def display_config(**kwargs):
    print("Configuration Settings:")
    print("-" * 25)
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

display_config(
    host="localhost",
    port=8080,
    debug=True,
    timeout=30
)
```

```python
# Example 3: Passing kwargs to another function (unpacking)
def inner_func(a, b, c):
    print(f"a={a}, b={b}, c={c}")

def outer_func(**kwargs):
    print(f"Received: {kwargs}")
    inner_func(**kwargs)  # Unpack dictionary as keyword arguments

outer_func(a=1, b=2, c=3)

# Example 4: Combining with regular parameters
def create_profile(name, age, **extra_info):
    profile = {"name": name, "age": age}
    profile.update(extra_info)  # Add any additional fields
    return profile

user = create_profile("Alice", 30, city="Toronto", role="Engineer", active=True)
print(user)
```

### Common Use Cases for **kwargs

1. **Wrapper functions** - Pass unknown arguments to wrapped functions
2. **Configuration options** - Accept flexible settings without defining each parameter
3. **API design** - Allow extensibility without breaking existing code
4. **Decorators** - Forward all arguments to the decorated function

## 5. Return Values

Functions can return single values, multiple values (as tuples), or `None` by default if no return statement is specified.

```python
# Single return value
def square(x):
    return x ** 2

result = square(5)
print(result)

# Multiple return values (returned as tuple)
def get_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

minimum, maximum, average = get_stats([1, 2, 3, 4, 5])
print(f"Min: {minimum}, Max: {maximum}, Avg: {average}")

# No return statement returns None
def say_hello(name):
    print(f"Hello, {name}")

result = say_hello("Alice")
print(f"Return value: {result}")
```

## 6. Lambda Functions

Lambda functions are small anonymous functions defined with the `lambda` keyword. They can have multiple arguments but only one expression.

```python
# Basic lambda
square = lambda x: x ** 2
print(square(5))

# Lambda with multiple arguments
add = lambda a, b: a + b
print(add(3, 4))

# Common use: with built-in functions
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
sorted_nums = sorted(numbers, key=lambda x: -x)  # Sort descending
print(sorted_nums)

# Lambda with filter and map
evens = list(filter(lambda x: x % 2 == 0, numbers))
squares = list(map(lambda x: x ** 2, numbers))
print(f"Evens: {evens}")
print(f"Squares: {squares}")
```

## 7. Scope and Closures

- **Local scope**: Variables inside a function
- **Global scope**: Variables at module level
- **nonlocal**: Access enclosing function's variables
- **Closure**: Inner function that remembers the enclosing scope

```python
# Local vs Global scope
global_var = "I'm global"

def demo_scope():
    local_var = "I'm local"
    print(global_var)  # Can read global
    print(local_var)

demo_scope()

# Modifying global variable
counter = 0

def increment():
    global counter
    counter += 1

increment()
print(f"Counter: {counter}")

# Closure example
def make_multiplier(n):
    def multiplier(x):
        return x * n  # n is "remembered" from enclosing scope
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))  # 10
print(triple(5))  # 15
```

## 8. Decorators

Decorators are functions that modify the behavior of other functions. They wrap a function, adding functionality before or after the original function runs.

```python
# Basic decorator
def log_calls(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@log_calls
def add(a, b):
    return a + b

add(3, 5)

# Decorator with timing
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(0.1)
    return "Done"

slow_function()
```

## 9. Recursive Functions

A recursive function calls itself to solve smaller instances of the same problem. Every recursive function needs a **base case** to stop the recursion.

```python
# Factorial: n! = n * (n-1)!
def factorial(n):
    if n <= 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case

print(f"5! = {factorial(5)}")

# Fibonacci: fib(n) = fib(n-1) + fib(n-2)
def fibonacci(n):
    if n <= 1:  # Base case
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"First 10 Fibonacci numbers: {[fibonacci(i) for i in range(10)]}")

# Sum of nested list (recursive)
def sum_nested(lst):
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)  # Recursively sum nested list
        else:
            total += item
    return total

nested = [1, [2, 3], [4, [5, 6]]]
print(f"Sum of {nested} = {sum_nested(nested)}")
```

## 10. Programming Exercises

Practice problems to reinforce function concepts.

```python
# Problem 1: Check if a number is prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(f"Is 17 prime? {is_prime(17)}")
print(f"Is 20 prime? {is_prime(20)}")
print(f"Primes up to 30: {[n for n in range(2, 31) if is_prime(n)]}")
```

```python
# Problem 2: Reverse a string without using slicing
def reverse_string(s):
    result = ""
    for char in s:
        result = char + result
    return result

print(reverse_string("hello"))  # olleh
print(reverse_string("Python"))  # nohtyP
```

```python
# Problem 3: Find the second largest number in a list
def second_largest(nums):
    if len(nums) < 2:
        return None
    first = second = float('-inf')
    for n in nums:
        if n > first:
            second = first
            first = n
        elif n > second and n != first:
            second = n
    return second if second != float('-inf') else None

print(second_largest([10, 5, 8, 20, 15]))  # 15
print(second_largest([1, 1, 1]))  # None (all same)
```

```python
# Problem 4: Check if a string is a palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print(is_palindrome("Racecar"))  # True
print(is_palindrome("A man a plan a canal Panama".replace(" ", "")))  # True
print(is_palindrome("Hello"))  # False
```

```python
# Problem 5: Count word frequency in a sentence
def word_frequency(sentence):
    words = sentence.lower().split()
    freq = []
    counted = []
    for word in words:
        if word not in counted:
            counted.append(word)
            freq.append((word, words.count(word)))
    return freq

text = "the quick brown fox jumps over the lazy dog the fox"
print(word_frequency(text))
```

```python
# Problem 6: Find common elements in two lists
def common_elements(list1, list2):
    result = []
    for item in list1:
        if item in list2 and item not in result:
            result.append(item)
    return result

a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
print(f"Common elements: {common_elements(a, b)}")  # [4, 5]
```

```python
# Problem 7: Flatten a nested list
def flatten(nested_list):
    result = []
    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

nested = [1, [2, 3], [4, [5, 6, [7, 8]]]]
print(f"Flattened: {flatten(nested)}")  # [1, 2, 3, 4, 5, 6, 7, 8]
```

```python
# Problem 8: FizzBuzz - classic interview problem
def fizzbuzz(n):
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(i)
    return result

print(fizzbuzz(15))
```

```python
# Problem 9: Two Sum - find indices of two numbers that add up to target
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

numbers = [2, 7, 11, 15]
print(two_sum(numbers, 9))  # [0, 1] because 2 + 7 = 9
```

```python
# Problem 10: Remove duplicates from a list while preserving order
def remove_duplicates(lst):
    seen = []
    result = []
    for item in lst:
        if item not in seen:
            seen.append(item)
            result.append(item)
    return result

nums = [1, 2, 2, 3, 4, 3, 5, 1, 6]
print(remove_duplicates(nums))  # [1, 2, 3, 4, 5, 6]
```
