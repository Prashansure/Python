## Section 1: Define Recursive Functions with Base and Recursive Cases

A recursive function must have:
1. **Base Case**: A condition that stops the recursion
2. **Recursive Case**: A call to the function itself with a simpler input

Without a base case, the function will call itself infinitely, causing a stack overflow.

Let's start with a simple example: summing a list of numbers.

```python
# Example 1: Sum of a list using recursion

def sum_list_recursive(numbers):
    """
    Sum all elements in a list using recursion.
    
    Base Case: Empty list returns 0
    Recursive Case: First element + sum of remaining elements
    """
    # Base case: empty list
    if len(numbers) == 0:
        return 0
    
    # Recursive case: first element + sum of rest
    return numbers[0] + sum_list_recursive(numbers[1:])

# Test the function
test_list = [1, 2, 3, 4, 5]
result = sum_list_recursive(test_list)
print(f"Sum of {test_list} = {result}")

# Another example
test_list_2 = [10, 20, 30, 40]
result_2 = sum_list_recursive(test_list_2)
print(f"Sum of {test_list_2} = {result_2}")

# Edge case: empty list
empty_result = sum_list_recursive([])
print(f"Sum of [] = {empty_result}")
```

## Section 2: Trace Recursion with Call Stack Visualization

Let's visualize how the call stack works by adding print statements to show when we enter and exit each recursive call.

```python
# Tracing call stack with print statements
depth = 0

def traced_sum(numbers):
    """
    Sum with call stack tracing.
    Shows entry and exit points for each recursive call.
    """
    global depth
    depth += 1
    current_depth = depth
    
    indent = "  " * (current_depth - 1)
    print(f"{indent}→ Enter: sum({numbers})")
    
    # Base case
    if len(numbers) == 0:
        print(f"{indent}← Base case reached, returning 0")
        depth -= 1
        return 0
    
    # Recursive case
    first = numbers[0]
    rest_sum = traced_sum(numbers[1:])
    result = first + rest_sum
    
    print(f"{indent}← Exit: sum({numbers}) = {first} + {rest_sum} = {result}")
    depth -= 1
    return result

# Test with a small list
print("=== Call Stack Trace ===")
result = traced_sum([1, 2, 3])
print(f"\nFinal result: {result}")
```

## Section 3: Compute Factorial and Fibonacci Recursively

### Factorial
Factorial of n (n!) = n × (n-1) × (n-2) × ... × 1
- Base case: 0! = 1, 1! = 1
- Recursive case: n! = n × (n-1)!

### Fibonacci
Fibonacci sequence: 0, 1, 1, 2, 3, 5, 8, 13, ...
- F(0) = 0, F(1) = 1
- F(n) = F(n-1) + F(n-2)

```python
def factorial(n):
    """
    Calculate factorial of n recursively.
    n! = n × (n-1) × (n-2) × ... × 1
    """
    # Base case
    if n == 0 or n == 1:
        return 1
    # Recursive case
    return n * factorial(n - 1)

# Test factorial
print("=== Factorial ===")
for i in range(6):
    result = factorial(i)
    print(f"{i}! = {result}")

print("\n" + "="*40 + "\n")

def fibonacci_naive(n):
    """
    Calculate nth Fibonacci number recursively (naive approach).
    F(0) = 0, F(1) = 1
    F(n) = F(n-1) + F(n-2)
    WARNING: Very slow for large n!
    """
    if n <= 1:
        return n
    return fibonacci_naive(n - 1) + fibonacci_naive(n - 2)

# Test Fibonacci - only small values due to exponential time complexity
print("=== Fibonacci (Naive - Slow) ===")
for i in range(10):
    result = fibonacci_naive(i)
    print(f"F({i}) = {result}")
```

```python
def fibonacci_memo(n, memo=None):
    """
    Calculate nth Fibonacci number with memoization (caching).
    Much faster than naive approach - O(n) instead of O(2^n)
    """
    if memo is None:
        memo = {}
    
    # Check if already computed
    if n in memo:
        return memo[n]
    
    # Base case
    if n <= 1:
        return n
    
    # Recursive case with memoization
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    return memo[n]

# Test optimized Fibonacci - can handle larger values
print("=== Fibonacci (Optimized with Memoization - Fast) ===")
for i in range(30):
    result = fibonacci_memo(i)
    print(f"F({i}) = {result}")

# Test with a very large value
large_fib = fibonacci_memo(100)
print(f"\nF(100) = {large_fib}")
```

## Section 4: Implement Divide and Conquer: Recursive Merge Sort

**Divide and Conquer Pattern:**
1. **Divide**: Split the array in half
2. **Conquer**: Recursively sort each half
3. **Combine**: Merge the two sorted halves

**Time Complexity**: O(n log n)
**Space Complexity**: O(n)

```python
def merge_sort_recursive(arr):
    """
    Sort array using recursive merge sort.
    
    Steps:
    1. Divide: Split array into two halves
    2. Conquer: Recursively sort each half
    3. Combine: Merge the sorted halves
    """
    # Base case: arrays with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr
    
    # Divide: split at midpoint
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    
    # Conquer: recursively sort each half
    left_sorted = merge_sort_recursive(left)
    right_sorted = merge_sort_recursive(right)
    
    # Combine: merge the sorted halves
    return merge(left_sorted, right_sorted)

def merge(left, right):
    """Merge two sorted arrays into one sorted array."""
    result = []
    i = j = 0
    
    # Compare elements from left and right, add smaller to result
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # Add remaining elements
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

# Test recursive merge sort
print("=== Recursive Merge Sort ===")
test_arrays = [
    [64, 34, 25, 12, 22, 11, 90],
    [5, 2, 8, 1, 9],
    [1],
    [],
    [100, 50, 75, 25, 90, 60]
]

for arr in test_arrays:
    sorted_arr = merge_sort_recursive(arr)
    print(f"Original: {arr}")
    print(f"Sorted:   {sorted_arr}\n")
```

## Section 5: Implement Divide and Conquer: Iterative Merge Sort

Instead of using recursion, we can implement merge sort iteratively by:
1. Starting with small subarrays (size 1)
2. Merging adjacent subarrays of increasing size
3. Continuing until the entire array is sorted

**Time Complexity**: O(n log n)
**Space Complexity**: O(n)
**Advantage**: Avoids recursion overhead and stack issues

```python
def merge_sort_iterative(arr):
    """
    Sort array using iterative merge sort (bottom-up approach).
    
    Strategy:
    1. Start with size 1 (each element is sorted)
    2. Merge adjacent subarrays of size 1 → size 2
    3. Merge adjacent subarrays of size 2 → size 4
    4. Continue doubling the size until entire array is sorted
    """
    n = len(arr)
    arr = arr.copy()  # Don't modify original
    
    # Start with subarray size of 1, then 2, 4, 8, etc.
    current_size = 1
    
    while current_size < n:
        # Pick starting point of left sub array to be merged
        left_start = 0
        
        while left_start < n:
            # Find ending point of left subarray
            left_end = min(left_start + current_size - 1, n - 1)
            
            # Find ending point of right subarray
            right_end = min(left_start + current_size * 2 - 1, n - 1)
            
            # Merge subarrays arr[left_start...left_end] and arr[left_end+1...right_end]
            if left_end < right_end:
                merge_in_place(arr, left_start, left_end, right_end)
            
            # Move to next pair of subarrays
            left_start += current_size * 2
        
        # Increase current size by 2 for next iteration
        current_size *= 2
    
    return arr

def merge_in_place(arr, left_start, left_end, right_end):
    """Merge two sorted subarrays in place."""
    # Create temporary arrays for left and right subarrays
    left = arr[left_start:left_end + 1]
    right = arr[left_end + 1:right_end + 1]
    
    i = j = 0
    k = left_start
    
    # Merge back into original array
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            arr[k] = left[i]
            i += 1
        else:
            arr[k] = right[j]
            j += 1
        k += 1
    
    # Copy remaining elements
    while i < len(left):
        arr[k] = left[i]
        i += 1
        k += 1
    
    while j < len(right):
        arr[k] = right[j]
        j += 1
        k += 1

# Test iterative merge sort
print("=== Iterative Merge Sort ===")
test_arrays = [
    [64, 34, 25, 12, 22, 11, 90],
    [5, 2, 8, 1, 9],
    [1],
    [],
    [100, 50, 75, 25, 90, 60]
]

for arr in test_arrays:
    sorted_arr = merge_sort_iterative(arr)
    print(f"Original: {arr}")
    print(f"Sorted:   {sorted_arr}\n")
```

## Section 6: Compare Recursion vs Iteration for Divide and Conquer

Let's benchmark both implementations and count comparisons to understand the performance characteristics.

```python
import time
import random

# Create test arrays of various sizes
test_sizes = [100, 500, 1000, 5000, 10000]
results = {"Size": [], "Recursive (ms)": [], "Iterative (ms)": [], "Speedup": []}

print("=== Performance Comparison: Recursive vs Iterative Merge Sort ===\n")

for size in test_sizes:
    # Generate random test array
    test_array = [random.randint(1, 10000) for _ in range(size)]
    
    # Test recursive implementation
    start_time = time.time()
    merge_sort_recursive(test_array.copy())
    recursive_time = (time.time() - start_time) * 1000  # Convert to milliseconds
    
    # Test iterative implementation
    start_time = time.time()
    merge_sort_iterative(test_array.copy())
    iterative_time = (time.time() - start_time) * 1000  # Convert to milliseconds
    
    # Calculate speedup
    speedup = recursive_time / iterative_time if iterative_time > 0 else 0
    
    results["Size"].append(size)
    results["Recursive (ms)"].append(f"{recursive_time:.3f}")
    results["Iterative (ms)"].append(f"{iterative_time:.3f}")
    results["Speedup"].append(f"{speedup:.2f}x")
    
    print(f"Size: {size:,}")
    print(f"  Recursive:  {recursive_time:.3f} ms")
    print(f"  Iterative:  {iterative_time:.3f} ms")
    print(f"  Speedup:    {speedup:.2f}x (iterative is faster)")
    print()

# Display summary table
print("\n=== Summary Table ===")
print(f"{'Size':<10} {'Recursive (ms)':<20} {'Iterative (ms)':<20} {'Speedup':<15}")
print("-" * 65)
for i in range(len(results["Size"])):
    print(f"{results['Size'][i]:<10} {results['Recursive (ms)'][i]:<20} "
          f"{results['Iterative (ms)'][i]:<20} {results['Speedup'][i]:<15}")
```

## Analysis and Key Takeaways

### Recursion vs Iteration Comparison

| Aspect | Theoretical | Practical |
|--------|-------------|-----------|
| **Time Complexity** | Both O(n log n) | Both equivalent |
| **Space Complexity** | Both O(n) | Both equivalent |
| **Actual Speed** | Same | Iterative often faster (less overhead) |
| **Code Clarity** | Recursive often clearer | Iterative more complex |
| **Stack Usage** | Recursive uses call stack | Iterative uses explicit loops |
| **Risk** | Stack overflow possible | No stack risk |

### When to Use Each

**Use Recursion When:**
- Problem has natural recursive structure (trees, graphs, divide-and-conquer)
- Code clarity and simplicity are priorities
- Input size is guaranteed to be manageable
- You can use memoization for efficiency

**Use Iteration When:**
- Working with large datasets (avoid stack issues)
- Performance is critical
- Preventing stack overhead is important
- Problem is naturally sequential

### Key Lessons from Recursion

1. **Always define a base case** - prevents infinite recursion
2. **Progress toward base case** - ensure each call moves closer to the solution
3. **Use memoization** - cache results to avoid recomputation (Fibonacci example)
4. **Watch recursion depth** - deep recursion risks stack overflow
5. **Consider tail recursion** - some languages optimize this
6. **Combine approaches** - don't hesitate to mix recursion and iteration

### Divide and Conquer Strategy

The divide-and-conquer approach is powerful for:
- **Sorting**: Merge Sort, Quick Sort
- **Searching**: Binary Search
- **Mathematical**: Power calculation, Matrix multiplication
- **Array problems**: Finding closest pair, maximum subarray

Both recursive and iterative implementations of divide-and-conquer achieve the same O(n log n) or better time complexity, but iterative versions often have lower overhead.
