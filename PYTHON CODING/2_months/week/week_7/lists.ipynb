##Using [] to create a list

```python
my_list = []
my_list = [1, 2]
my_list = [1, 2, "siri", 4.2, 5]
```

##Using range() to create a list

```python
my_list = list(range(10))
print(my_list)
```

```python
my_list = list(range(0,10,2))
print(my_list)
```

##Accessing a list

```python
#indexing
print(my_list[0])
```

```python
#using loops
for i in my_list:
  print(i)
```

```python
#using slicing
print(my_list[0:2])
print(my_list[2:])
print(my_list[:3])
print(my_list[:])
print(my_list[-1])
```

##Updating lists

```python
  my_list[0] = 10
  print(my_list)
  my_list[1:3] = [11, 12]
  print(my_list)
  del my_list[0]
  print(my_list)
  my_list.append(13)
  print(my_list)
```

```python
empty_list = []

empty_list.append("not")
empty_list.append("so")
empty_list.append("empty")
empty_list.append("anymore")

print(empty_list)
```

#Nested Lists

```python
a=[10,20,30]
b=[45,65,a]
print(b) # display [ 45, 65, [ 10, 20, 30 ] ]
print(b[1]) # display 65
print(b[2]) # display [ 10, 20, 30 ]
print(b[2][0]) # display 10
print(b[2][1]) # display 20
print(b[2][2]) # display 30
for x in b[2]:
    print(x) # display 10 20 30
```

##Nested Lists as Matrices:

```python
mat=[[1,2,3],[4,5,6],[7,8,9]]
m=len(mat)
n=len(mat[0])
for i in range(0,m):
 for j in range(0,n):
    print(mat[i][j])
 print("")
print("")
```
