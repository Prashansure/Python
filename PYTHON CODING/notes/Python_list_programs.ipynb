## 1. Find Elements Occurring More Than Once (Duplicates)

```python
# Find elements occurring more than once
nums = [1, 2, 3, 2, 4, 3, 5, 6, 3]

# Using list comprehension and count()
duplicates = []
for x in nums:
    if nums.count(x) > 1 and x not in duplicates:
        duplicates.append(x)
print("Duplicates:", duplicates)
```

## 2. Remove Duplicates While Preserving Order

```python
# Remove duplicates while preserving order
nums = [1, 2, 3, 2, 4, 3, 5]

# Using loop to track seen elements
seen = []
for x in nums:
    if x not in seen:
        seen.append(x)
print("Unique (ordered):", seen)
```

## 3. Find Second Largest Element

```python
# Find second largest element
nums = [10, 20, 4, 45, 99, 99]

# Method 1: Sort and get unique second largest
unique_sorted = sorted(set(nums), reverse=True)
second_largest = unique_sorted[1] if len(unique_sorted) > 1 else None
print("Second largest:", second_largest)

# Method 2: Single pass (O(n))
first = second = float()
for n in nums:
    if n > first:
        second = first
        first = n
    elif n > second and n != first:
        second = n
print("Second largest (single pass):", second)
```

## 4. Rotate List by K Positions

```python
# Rotate list by k positions
nums = [1, 2, 3, 4, 5]
k = 2

# Right rotation using slicing
rotated_right = nums[-k:] + nums[:-k]
print(f"Right rotate by {k}:", rotated_right)

# Left rotation using slicing
rotated_left = nums[k:] + nums[:k]
print(f"Left rotate by {k}:", rotated_left)
```

## 5. Find Intersection of Two Lists

```python
# Find intersection of two lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

# Using list comprehension
intersection = [x for x in list1 if x in list2]
print("Intersection:", intersection)
```

## 6. Flatten a Nested List

```python
# Flatten a nested list
nested = [[1, 2], [3, 4], [5, 6]]

# Method 1: List comprehension (one level)
flat = [item for sublist in nested for item in sublist]
print("Flattened:", flat)

# Method 2: Recursive (handles deep nesting)
def flatten(lst):
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

deeply_nested = [1, [2, [3, [4, 5]]]]
print("Deep flatten:", flatten(deeply_nested))
```

## 7. Find Missing Number in Range

```python
# Find missing number in a range [1, n]
nums = [1, 2, 4, 5, 6]  # Missing 3
n = 6

# Using sum formula
expected_sum = n * (n + 1) // 2
actual_sum = sum(nums)
missing = expected_sum - actual_sum
print("Missing number:", missing)

# Using list comprehension
missing_nums = [x for x in range(1, n + 1) if x not in nums]
print("Missing (list):", missing_nums)
```

## 8. Two Sum - Find Pairs That Sum to Target

```python
# Two Sum - Find indices of two numbers that add up to target
nums = [2, 7, 11, 15]
target = 9

# Using nested loops (O(n²))
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

result = two_sum(nums, target)
print(f"Indices that sum to {target}:", result)
```

## 9. Merge Two Sorted Lists

```python
# Merge two sorted lists
list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]

# Method 1: Simple (but O(n log n))
merged = sorted(list1 + list2)
print("Merged (sorted):", merged)

# Method 2: Two pointers (O(n))
def merge_sorted(l1, l2):
    result = []
    i = j = 0
    while i < len(l1) and j < len(l2):
        if l1[i] <= l2[j]:
            result.append(l1[i])
            i += 1
        else:
            result.append(l2[j])
            j += 1
    result.extend(l1[i:])
    result.extend(l2[j:])
    return result

print("Merged (two pointers):", merge_sorted(list1, list2))
```

## 10. Find Most Frequent Element

```python
# Find most frequent element
nums = [1, 3, 2, 3, 4, 3, 2, 3]

# Method 1: Using Counter
from collections import Counter
most_common = Counter(nums).most_common(1)[0]
print(f"Most frequent: {most_common[0]} (count: {most_common[1]})")

# Method 2: Using max with key
most_freq = max(set(nums), key=nums.count)
print("Most frequent:", most_freq)
```
