## 1. Adding Elements - `append()`, `insert()`, `extend()`

These methods allow you to add elements to a list:
- **`append(x)`**: Adds element `x` to the end of the list
- **`insert(i, x)`**: Inserts element `x` at position `i`
- **`extend(iterable)`**: Adds all elements from an iterable to the end of the list

# Python List Methods 

Demonstrates all methods associated with Python Lists with practical examples.

**Table of Contents:**
1. Adding Elements - `append()`, `insert()`, `extend()`
2. Removing Elements - `remove()`, `pop()`, `clear()`
3. Searching and Counting - `index()`, `count()`
4. Sorting and Reversing - `sort()`, `reverse()`
5. Copying Lists - `copy()`

```python
# append(x) - Adds a single element to the end
colors = ["red", "green", "blue"]
print("Original:", colors)

colors.append("yellow")
print("After append('yellow'):", colors)

# Note: Appending a list adds it as a nested element
colors.append(["purple", "orange"])
print("After append(list):", colors)
```

```python
# insert(i, x) - Inserts element at a specific position
animals = ["cat", "dog", "bird"]
print("Original:", animals)

animals.insert(1, "rabbit")  # Insert at index 1
print("After insert(1, 'rabbit'):", animals)

animals.insert(0, "fish")    # Insert at beginning
print("After insert(0, 'fish'):", animals)
```

```python
# extend(iterable) - Adds all elements from an iterable
numbers = [1, 2, 3]
print("Original:", numbers)

numbers.extend([4, 5, 6])
print("After extend([4, 5, 6]):", numbers)

# Works with any iterable (tuple, string, range)
letters = ["a", "b"]
letters.extend("xyz")  # Adds each character
print("After extend('xyz'):", letters)
```

## 2. Removing Elements - `remove()`, `pop()`, `clear()`

These methods allow you to remove elements from a list:
- **`remove(x)`**: Removes the first occurrence of value `x`
- **`pop([i])`**: Removes and returns the element at position `i` (default: last element)
- **`clear()`**: Removes all elements from the list

```python
# remove(x) - Removes the first occurrence of a value
fruits = ["apple", "banana", "cherry", "banana"]
print("Original:", fruits)

fruits.remove("banana")  # Removes only the first 'banana'
print("After remove('banana'):", fruits)

# Safe removal - check before removing to avoid ValueError
if "cherry" in fruits:
    fruits.remove("cherry")
    print("After remove('cherry'):", fruits)
```

```python
# pop([i]) - Removes and returns element by index
numbers = [10, 20, 30, 40, 50]
print("Original:", numbers)

last = numbers.pop()       # Pop last element (default)
print(f"Popped: {last}, List now: {numbers}")

second = numbers.pop(1)    # Pop at specific index
print(f"Popped index 1: {second}, List now: {numbers}")

first = numbers.pop(0)     # Pop first element
print(f"Popped first: {first}, List now: {numbers}")
```

```python
# clear() - Removes all elements from the list
items = ["a", "b", "c", "d"]
print("Before clear():", items)

items.clear()
print("After clear():", items)  # Empty list, but still exists
```

## 3. Searching and Counting - `index()`, `count()`

These methods help you find elements in a list:
- **`index(x[, start[, end]])`**: Returns the index of the first occurrence of `x`
- **`count(x)`**: Returns the number of occurrences of `x` in the list

```python
# index(x) - Returns the index of the first occurrence
fruits = ["apple", "banana", "cherry", "banana"]
print("List:", fruits)

idx = fruits.index("cherry")
print(f"Index of 'cherry': {idx}")

idx = fruits.index("banana")
print(f"Index of first 'banana': {idx}")

# Search within a range: index(x, start, end)
idx = fruits.index("banana", 2)  # Start from index 2
print(f"Index of 'banana' from index 2: {idx}")

# Note: Raises ValueError if element not found
```

```python
# count(x) - Returns the number of occurrences
numbers = [1, 2, 3, 2, 4, 2, 5]
print("List:", numbers)
print(f"Count of 2: {numbers.count(2)}")
print(f"Count of 10: {numbers.count(10)}")  # Returns 0 if not found

# Note: 1, True, and 1.0 are considered equal in Python
mixed = [1, True, 1.0, "1"]
print(f"Count of 1 in {mixed}: {mixed.count(1)}")
```

## 4. Sorting and Reversing - `sort()`, `reverse()`

These methods modify the order of elements in a list:
- **`sort(key=None, reverse=False)`**: Sorts the list in place (ascending by default)
- **`reverse()`**: Reverses the list in place

```python
# sort() - Sorts the list in place
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
print("Original:", numbers)

numbers.sort()  # Ascending (default)
print("Ascending:", numbers)

numbers.sort(reverse=True)  # Descending
print("Descending:", numbers)

# Sorting strings (case-sensitive by default)
words = ["Banana", "apple","cherry"]
words.sort()
print(words)
words.sort(key=str.upper)  # Case-insensitive sort
print("Case-insensitive:", words)
```

```python
# sort() with custom key function
words = ["python", "is", "awesome", "fun"]
words.sort(key=len)  # Sort by string length
print("Sorted by length:", words)
```

```python
# reverse() - Reverses the list in place
numbers = [1, 2, 3, 4, 5]
print("Original:", numbers)

numbers.reverse()
print("After reverse():", numbers)

# Difference: reverse() vs [::-1]
original = [1, 2, 3, 4, 5]
original.reverse()         # Modifies in place, returns None
reversed_copy = [1, 2, 3, 4, 5][::-1]  # Creates new list
print("[::-1] creates new list:", reversed_copy)

# reversed(): returns an iterator that accesses elements in reverse order without modifying the original sequence.

lst = [1, 2, 3, 4]
rev = reversed(lst)

print("reversed list",list(rev))      # [4, 3, 2, 1] 
print("original list", lst)
# Works with Lists, tuples, strings, ranges, any sequence
```

## 5. Copying Lists - `copy()`

The `copy()` method creates a shallow copy of the list:
- **`copy()`**: Returns a shallow copy of the list

**Important:** A shallow copy creates a new list but references the same nested objects.

```python
# copy() - Creates a shallow copy of the list
original = [1, 2, 3, 4, 5]
copied = original.copy()

copied.append(6)
print("Original:", original)  # Unchanged
print("Copy:", copied)
```

```python
# Shallow vs Deep Copy - Important for nested lists!
import copy

# Shallow copy - nested objects are shared. Faster and less memory
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow[0][0] = 999
print("Original after shallow edit:", original)  # Also modified!

# Deep copy - completely independent, Safe to modify
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)
deep[0][0] = 999
print("after deep edit:", deep)
print("Original after deep edit:", original)  # Unchanged

```

## Summary of List Methods

| Method | Description | Returns |
|--------|-------------|---------|
| `append(x)` | Adds element x to end | None |
| `extend(iter)` | Adds all elements from iterable | None |
| `insert(i, x)` | Inserts x at position i | None |
| `remove(x)` | Removes first occurrence of x | None |
| `pop([i])` | Removes and returns element at i | Element |
| `clear()` | Removes all elements | None |
| `index(x)` | Returns index of first x | int |
| `count(x)` | Returns count of x occurrences | int |
| `sort()` | Sorts list in place | None |
| `reverse()` | Reverses list in place | None |
| `copy()` | Returns shallow copy | list |

**Note:** Methods that modify the list in place return `None`.
