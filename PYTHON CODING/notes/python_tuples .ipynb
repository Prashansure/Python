## 1. Creating Tuples

Tuples are ordered, immutable sequences. They are defined using parentheses `()` or just commas.

```python
# Different ways to create tuples
empty_tuple = ()
single_element = (42,)  # Note: comma is required for single element
numbers = (1, 2, 3, 4, 5)
mixed = (1, "hello", 3.14, True)

# Without parentheses (tuple packing)
coordinates = 10, 20, 30

# Using tuple() constructor
from_list = tuple([1, 2, 3])
from_string = tuple("hello")

print(f"Empty: {empty_tuple}")
print(f"Single element: {single_element}, type: {type(single_element)}")
print(f"Numbers: {numbers}")
print(f"Mixed types: {mixed}")
print(f"Coordinates: {coordinates}")
print(f"From list: {from_list}")
print(f"From string: {from_string}")
```

### Understanding Tuple Packing

**Tuple packing** is when Python automatically combines comma-separated values into a tuple. The parentheses are optional—it's the **comma** that creates the tuple.

| Syntax | Result | Type |
|--------|--------|------|
| `x = 10, 20, 30` | `(10, 20, 30)` | tuple |
| `x = (10, 20, 30)` | `(10, 20, 30)` | tuple |
| `x = (10)` | `10` | int ⚠️ |
| `x = (10,)` | `(10,)` | tuple ✓ |
| `x = 10,` | `(10,)` | tuple ✓ |

```python
# Tuple packing examples - parentheses are OPTIONAL

# With parentheses
point1 = (10, 20, 30)

# Without parentheses (same result!)
point2 = 10, 20, 30

print(f"With parentheses: {point1}, type: {type(point1)}")
print(f"Without parentheses: {point2}, type: {type(point2)}")
print(f"Are they equal? {point1 == point2}")

# CAUTION: Single element tuple gotcha
not_a_tuple = (42)    # This is just an integer!
is_a_tuple = (42,)    # Comma makes it a tuple
also_tuple = 42,      # Works without parentheses too

print(f"\n(42) is type: {type(not_a_tuple)}")
print(f"(42,) is type: {type(is_a_tuple)}")
print(f"42, is type: {type(also_tuple)}")
```

```python
# Common uses of tuple packing (without parentheses)

# 1. Function returns - very common pattern
def get_coordinates():
    return 100, 200, 300  # Returns tuple without parentheses

coords = get_coordinates()
print(f"Returned: {coords}")

# 2. Variable swapping - elegant one-liner
a, b = 10, 20
print(f"Before swap: a={a}, b={b}")
a, b = b, a  # Packs (b, a) then unpacks to a, b
print(f"After swap: a={a}, b={b}")

# 3. Multiple assignment
name, age, city = "Alice", 30, "Toronto"
print(f"{name} is {age} years old from {city}")

# 4. Enumerate returns tuples (packed implicitly)
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

## 2. Accessing Tuple Elements

Access elements using indexing (0-based) and slicing, just like lists.

```python
fruits = ("apple", "banana", "cherry", "date", "elderberry")

# Positive indexing (0-based)
print(f"First element: {fruits[0]}")
print(f"Third element: {fruits[2]}")

# Negative indexing (from end)
print(f"Last element: {fruits[-1]}")
print(f"Second to last: {fruits[-2]}")

# Slicing [start:stop:step]
print(f"First three: {fruits[:3]}")
print(f"Last two: {fruits[-2:]}")
print(f"Every other: {fruits[::2]}")
print(f"Reversed: {fruits[::-1]}")

# Check membership
print(f"'banana' in tuple: {'banana' in fruits}")
print(f"Length: {len(fruits)}")
```

## 3. Tuple Immutability

Tuples cannot be modified after creation. This makes them hashable and safer for data that shouldn't change.

```python
my_tuple = (1, 2, 3)

# This will raise an error - uncomment to see
# my_tuple[0] = 10  # TypeError: 'tuple' object does not support item assignment

# Workaround: Convert to list, modify, convert back
my_list = list(my_tuple)
my_list[0] = 10
my_tuple = tuple(my_list)
print(f"Modified tuple: {my_tuple}")

# IMPORTANT: Mutable objects inside tuples CAN be modified
nested = ([1, 2, 3], [4, 5, 6])
nested[0].append(99)  # This works!
print(f"Nested with modified list: {nested}")

# Tuple concatenation creates a NEW tuple
original = (1, 2, 3)
new_tuple = original + (4, 5)
print(f"Original: {original}")
print(f"New tuple: {new_tuple}")
```

## 4. Tuple Methods

Tuples have only two built-in methods: `count()` and `index()`.

```python
numbers = (1, 2, 2, 3, 2, 4, 5, 2)

# count() - returns number of occurrences
count_of_2 = numbers.count(2)
print(f"Number 2 appears {count_of_2} times")

# index() - returns first index of value
first_index = numbers.index(2)
print(f"First occurrence of 2 at index: {first_index}")

# index() with start and end parameters
second_index = numbers.index(2, first_index + 1)  # Start searching after first occurrence
print(f"Second occurrence of 2 at index: {second_index}")

# Other useful operations (not methods, but built-in functions)
print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")
print(f"Sum: {sum(numbers)}")
print(f"Sorted (returns list): {sorted(numbers)}")
```

## 5. Tuple Packing and Unpacking

Packing combines values into a tuple; unpacking extracts values from a tuple into variables.

```python
# Tuple packing
person = "Alice", 30, "Engineer"  # Parentheses optional
print(f"Packed: {person}")

# Tuple unpacking
name, age, role = person
print(f"Name: {name}, Age: {age}, Role: {role}")

# Swapping variables (elegant Python way)
a, b = 10, 20
a, b = b, a
print(f"After swap: a={a}, b={b}")

# Extended unpacking with *
first, *middle, last = (1, 2, 3, 4, 5)
print(f"First: {first}, Middle: {middle}, Last: {last}")

# Ignoring values with _
x, _, z = (10, 20, 30)
print(f"x={x}, z={z} (ignored middle)")
x, z, _ = (10, 20, 30)
print(f"x={x}, z={z} (ignored last)")

# Unpacking in function returns
def get_user():
    return "Bob", 25, "Developer"

name, age, role = get_user()
print(f"User: {name}, {age}, {role}")
```

## 6. Tuples vs Lists

| Feature | Tuple | List |
|---------|-------|------|
| Syntax | `(1, 2, 3)` | `[1, 2, 3]` |
| Mutability | Immutable | Mutable |
| Hashable | Yes (if all elements hashable) | No |
| Performance | Faster | Slower |
| Memory | Less | More |
| Use case | Fixed data | Dynamic data |

```python
import sys

# Memory comparison
my_list = [1, 2, 3, 4, 5]
my_tuple = (1, 2, 3, 4, 5)

print(f"List size: {sys.getsizeof(my_list)} bytes")
print(f"Tuple size: {sys.getsizeof(my_tuple)} bytes")

# Performance comparison
import timeit

list_time = timeit.timeit('[1, 2, 3, 4, 5]', number=1000000)
tuple_time = timeit.timeit('(1, 2, 3, 4, 5)', number=1000000)

print(f"\nCreating 1M times:")
print(f"List: {list_time:.4f} seconds")
print(f"Tuple: {tuple_time:.4f} seconds")
print(f"Tuple is {list_time/tuple_time:.2f}x faster")
```

## 7. Named Tuples

Named tuples add readable field names while keeping tuple benefits. Great for structured data without creating full classes.

```python
from collections import namedtuple

# Define a named tuple type
Person = namedtuple('Person', ['name', 'age', 'city'])

# Create instances
alice = Person('Alice', 30, 'Toronto')
bob = Person(name='Bob', age=25, city='Vancouver')

# Access by name (readable)
print(f"{alice.name} is {alice.age} years old from {alice.city}")

# Still works like regular tuple
print(f"Index access: {alice[0]}")
print(f"Unpack: ", end="")
name, age, city = bob
print(f"{name}, {age}, {city}")

# Useful methods
print(f"\nAs dict: {alice._asdict()}")
print(f"Fields: {Person._fields}")

# Create from dict
data = {'name': 'Charlie', 'age': 35, 'city': 'Montreal'}
charlie = Person(**data)
print(f"From dict: {charlie}")
```

## 8. Tuples as Dictionary Keys

Since tuples are hashable (immutable), they can be used as dictionary keys—lists cannot.

```python
# Using tuples as dictionary keys (coordinates example)
locations = {
    (40.7128, -74.0060): "New York",
    (51.5074, -0.1278): "London",
    (35.6762, 139.6503): "Tokyo"
}

print(f"Location at (40.7128, -74.0060): {locations[(40.7128, -74.0060)]}")

# Practical use: counting pairs/combinations
word_pairs = [("hello", "world"), ("foo", "bar"), ("hello", "world")]
pair_count = {}
for pair in word_pairs:
    pair_count[pair] = pair_count.get(pair, 0) + 1
print(f"Pair counts: {pair_count}")

# Multi-dimensional lookup (like a sparse matrix)
matrix = {}
matrix[(0, 0)] = 1
matrix[(1, 2)] = 5
matrix[(3, 1)] = 8
print(f"Matrix: {matrix}")

# This would fail with lists - uncomment to see error
# bad_dict = {[1, 2]: "value"}  # TypeError: unhashable type: 'list'
```

## 9. Programming Exercises

Practice problems to reinforce tuple concepts.

```python
# Problem 1: Find min and max in a tuple
def min_max(t):
    return (min(t), max(t))

numbers = (5, 2, 9, 1, 7, 3)
result = min_max(numbers)
print(f"Min and Max of {numbers}: {result}")
```

```python
# Problem 2: Count elements in a tuple of tuples
def count_all_elements(nested_tuple):
    count = 0
    for item in nested_tuple:
        if isinstance(item, tuple):
            count += count_all_elements(item)
        else:
            count += 1
    return count

data = ((1, 2), (3, 4, 5), ((6, 7), 8))
print(f"Total elements in {data}: {count_all_elements(data)}")
```

```python
# Problem 3: Sort a list of tuples by second element
def sort_by_second(tuple_list):
    return sorted(tuple_list, key=lambda x: x[1])

students = [("Alice", 85), ("Bob", 72), ("Charlie", 91), ("Diana", 68)]
sorted_students = sort_by_second(students)
print(f"Sorted by score: {sorted_students}")
```

```python
# Problem 4: Merge two tuples alternately
def merge_alternate(t1, t2):
    result = []
    min_len = min(len(t1), len(t2))
    for i in range(min_len):
        result.append(t1[i])
        result.append(t2[i])
    result.extend(t1[min_len:])
    result.extend(t2[min_len:])
    return tuple(result)

tuple1 = (1, 3, 5)
tuple2 = (2, 4, 6, 8, 10)
print(f"Merged: {merge_alternate(tuple1, tuple2)}")
```

```python
# Problem 5: Find common elements between two tuples
def common_elements(t1, t2):
    result = []
    for item in t1:
        if item in t2 and item not in result:
            result.append(item)
    return tuple(result)

a = (1, 2, 3, 4, 5)
b = (4, 5, 6, 7, 8)
print(f"Common elements: {common_elements(a, b)}")
```

```python
# Problem 6: Reverse a tuple
def reverse_tuple(t):
    return t[::-1]

original = (1, 2, 3, 4, 5)
print(f"Original: {original}")
print(f"Reversed: {reverse_tuple(original)}")
```

```python
# Problem 7: Group elements by type
def group_by_type(t):
    integers = []
    strings = []
    floats = []
    others = []
    
    for item in t:
        if isinstance(item, int) and not isinstance(item, bool):
            integers.append(item)
        elif isinstance(item, str):
            strings.append(item)
        elif isinstance(item, float):
            floats.append(item)
        else:
            others.append(item)
    
    return (tuple(integers), tuple(strings), tuple(floats), tuple(others))

mixed = (1, "hello", 3.14, 2, "world", True, 5.0, None)
ints, strs, flts, other = group_by_type(mixed)
print(f"Integers: {ints}")
print(f"Strings: {strs}")
print(f"Floats: {flts}")
print(f"Others: {other}")
```

```python
# Problem 8: Check if tuple is a subset of another
def is_subset(t1, t2):
    for item in t1:
        if item not in t2:
            return False
    return True

small = (1, 2, 3)
large = (1, 2, 3, 4, 5)
print(f"{small} is subset of {large}: {is_subset(small, large)}")
print(f"{large} is subset of {small}: {is_subset(large, small)}")
```

```python
# Problem 9: Flatten nested tuples
def flatten_tuple(nested):
    result = []
    for item in nested:
        if isinstance(item, tuple):
            result.extend(flatten_tuple(item))
        else:
            result.append(item)
    return tuple(result)

nested = ((1, 2), (3, (4, 5)), (6, (7, (8, 9))))
print(f"Nested: {nested}")
print(f"Flattened: {flatten_tuple(nested)}")
```

```python
# Problem 10: Find the most frequent element in a tuple
def most_frequent(t):
    max_count = 0
    most_common = None
    counted = []
    
    for item in t:
        if item not in counted:
            counted.append(item)
            count = t.count(item)
            if count > max_count:
                max_count = count
                most_common = item
    
    return (most_common, max_count)

data = (1, 2, 2, 3, 3, 3, 4, 4, 4, 4)
element, count = most_frequent(data)
print(f"Most frequent: {element} (appears {count} times)")
```
