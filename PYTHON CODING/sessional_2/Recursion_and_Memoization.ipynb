## Linear Recursion

A function calls itself exactly once per invocation. This is the simplest form of recursion.
It maps cleanly to problems that reduce in one step, like factorial or list processing.
The base case stops the call chain and returns the final result.

```python
# Linear recursion example: factorial
# Calculates factorial (n!) using one recursive call per function invocation.
# The base case is n <= 1, and each recursive step multiplies n by the factorial of (n-1).
def factorial(n):
    """Compute n! using linear recursion."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print("factorial(5) =", factorial(5))
```

```python
# Linear recursion example: power
# Computes base^exp by recursively multiplying the base (exp) times.
# Base case: any number to the power 0 equals 1.
# Each recursion reduces the exponent by 1, multiplying by base each time.
def power(base, exp):
    """Compute base^exp using linear recursion."""
    if exp == 0:
        return 1

    return base * power(base, exp - 1)print("2^5 =", power(2, 5))
```

```python
# Linear recursion example: string reversal
# Reverses a string by extracting the first character and recursing on the rest.
# Base case: empty string returns empty string.
# Each recursive call returns the reversed tail concatenated with the first character.
# Note: This approach creates many intermediate strings, so it's inefficient for long strings.
def reverse_string(text):
    """Reverse a string using linear recursion."""
    if text == "":
        return text

    return reverse_string(text[1:]) + text[0]print("reverse_string('hello') =", reverse_string("hello"))
```

## Binary Recursion

A function calls itself twice per invocation. This often leads to exponential time without optimization.
Binary recursion appears in divide-and-conquer and combinatorial problems.
It is a common target for memoization because overlapping subproblems are typical.

```python
# Binary recursion example: naive Fibonacci
# Computes Fibonacci numbers by making two recursive calls per function (binary recursion).
# This demonstrates the inefficiency of naive recursion: many subproblems are solved multiple times.
# For example, fib(5) computes fib(3) twice, fib(2) three times, etc.
# Observe the exponential growth in function calls—this is why memoization is essential.
import time

call_count = 0

def fibonacci_naive(n):
    """Compute Fibonacci with binary recursion (no memoization)."""
    global call_count
    call_count += 1
    if n <= 1:
        return n
    return fibonacci_naive(n - 1) + fibonacci_naive(n - 2)

n = 30
call_count = 0
start = time.time()
result = fibonacci_naive(n)
elapsed = time.time() - start


print(f"fibonacci({n}) = {result}")
print("time (seconds):", round(elapsed, 4))
print("calls:", call_count)
```

```python
# Binary recursion call tree visualization for a small n
# Prints the call tree structure with indentation to show recursion depth and branching.
# This helps visualize how fib(n) recursively calls fib(n-1) and fib(n-2), creating a binary tree.
# The number of leaf nodes grows exponentially, illustrating why naive Fibonacci is inefficient.
# Use small values of n (like 5-6) to keep output readable.

def fibonacci_trace(n, indent=0):
    print("  " * indent + f"fib({n})")
    if n <= 1:
        return n

    return fibonacci_trace(n - 1, indent + 1) + fibonacci_trace(n - 2, indent + 1)
fibonacci_trace(5)
```

## Nested Recursion

A recursive call is nested inside another recursive call.
This is one of the more complex forms of recursion and can grow very quickly.
Use small inputs when experimenting to avoid deep recursion or long runtimes.

```python
# Nested recursion example: Ackermann function (small inputs)
# The Ackermann function is famous for its explosive growth: recursive calls are nested within other calls.
# The pattern: A(0,n) = n+1, A(m,0) = A(m-1,1), A(m,n) = A(m-1, A(m,n-1)).
# Even small inputs grow rapidly (A(3,3) = 61). A(4,2) and beyond are enormous.
# Only test with m < 4 and n < 4 to avoid stack overflow. Shows limits of naive recursion.

def ackermann(m, n):
    if m == 0:
        return n + 1
    if n == 0:
        return ackermann(m - 1, 1)
    return ackermann(m - 1, ackermann(m, n - 1))

for m in range(3):

    for n in range(4):    
        print()
        print(f"A({m},{n}) = {ackermann(m, n)}", end="  ")
```

## Tree Recursion

A function calls itself multiple times to explore different branches, common in tree/graph problems.
This structure mirrors trees directly: each node spawns recursive work for its children.
It is frequently used in traversals, aggregation, and search algorithms.

```python
# Tree recursion example: sum and height of a binary tree
# Tree recursion solves subproblems on both left and right children, common in tree/graph algorithms.
# sum_tree: recursively sums all node values by adding root + left subtree sum + right subtree sum.
# tree_height: finds the maximum height by taking 1 + max(left height, right height).
# These patterns generalize to many tree problems: searching, validating, transforming trees.

class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

def sum_tree(node):
    if not node:
        return 0
    return node.value + sum_tree(node.left) + sum_tree(node.right)

def tree_height(node):
    if not node:
        return -1
    return 1 + max(tree_height(node.left), tree_height(node.right))

# Build a sample tree
#       1
#      / \
#     2   3
#    / \
#   4   5

root = TreeNode(1,
    TreeNode(2, TreeNode(4), TreeNode(5)),
    TreeNode(3)
)

print("tree_height =", tree_height(root))
print("sum_tree =", sum_tree(root))
```

## Tail vs Non-Tail Recursion

Tail recursion means the recursive call is the last operation. Non-tail recursion does additional work after the recursive call returns.
Tail-recursive functions are easier to optimize in some languages, though Python does not apply tail-call optimization.
Understanding the difference helps you spot where stack growth or extra work occurs.

```python
# Non-tail recursion example: sum 1..n
# In non-tail recursion, the recursive call is NOT the last operation.
# Here, sum_non_tail(n-1) is called, then n is added to the result—work happens after the return.
# Compare this with the tail-recursive version next to see the structural difference.

def sum_non_tail(n):
    if n == 0:
        return 0

    return n + sum_non_tail(n - 1)print("sum_non_tail(5) =", sum_non_tail(5))
```

```python
# Tail recursion example: sum 1..n with accumulator
# Tail recursion has the recursive call as the LAST operation (no work after return).
# The accumulator parameter carries the running sum, so we only need to return its value.
# Compiler optimizations (in languages like Scheme) can eliminate tail frames, saving stack space.
# Python doesn't optimize tail calls, but the pattern is still useful for clarity and potential porting.

def sum_tail(n, acc=0):
    if n == 0:
        return acc

    return sum_tail(n - 1, acc + n)print("sum_tail(5) =", sum_tail(5))
```

```python
# Non-tail vs tail recursion: factorial comparison
# Both functions compute n!, but structured differently:
# factorial_non_tail: multiplies n by the result of factorial(n-1)—work happens after the call.
# factorial_tail: passes the running product (acc) and returns it directly—work happens before the call.
# In tail-optimizing compilers, tail recursion uses constant stack space; non-tail grows with n.
# Python doesn't optimize tail calls, but the pattern is educational and useful in other languages.

def factorial_non_tail(n):
    if n <= 1:
        return 1
    return n * factorial_non_tail(n - 1)

def factorial_tail(n, acc=1):
    if n <= 1:
        return acc

    return factorial_tail(n - 1, acc * n)print("factorial_tail(5) =", factorial_tail(5))

print("factorial_non_tail(5) =", factorial_non_tail(5))
```

## Memoization

Memoization caches results of recursive calls so repeated subproblems do not get recomputed.
It is one of the most effective ways to speed up recursive algorithms with overlapping subproblems.
The next cells show three different ways to apply memoization in Python.

```python
# Memoization method 1: dictionary cache
# Memoization stores computed results to avoid redundant recursive calls.
# Here, a dictionary 'memo' stores fib(n) values. Before computing fib(n), we check if it's cached.
# The dramatic speedup shows memoization's power: naive fib(30) makes millions of calls,
# while memoized fib(30) computes each value only once for significant performance gains.
import time

def fibonacci_no_memo(n):
    if n <= 1:
        return n
    return fibonacci_no_memo(n - 1) + fibonacci_no_memo(n - 2)

def fibonacci_with_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci_with_memo(n - 1, memo) + fibonacci_with_memo(n - 2, memo)
    return memo[n]

n = 30
start = time.time()
result_no_memo = fibonacci_no_memo(n)
time_no_memo = time.time() - start

start = time.time()
result_with_memo = fibonacci_with_memo(n)
time_with_memo = time.time() - start

print(f"fibonacci({n}) without memo = {result_no_memo}")

print("time without memo (seconds):", round(time_no_memo, 4))
print("time with memo (seconds):", round(time_with_memo, 6))
print(f"fibonacci({n}) with memo = {result_with_memo}")
```

```python
# Memoization method 2: functools.lru_cache
# Python's @lru_cache decorator automates memoization without manual dictionary management.
# lru_cache (Least Recently Used) tracks function results and limits cache size (maxsize=None = unlimited).
# cache_info() shows hits, misses, and cache size; cache_clear() empties the cache for reuse.
# This is the cleaner, Pythonic approach for memoization in production code.
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci_lru(n):
    if n <= 1:
        return n
    return fibonacci_lru(n - 1) + fibonacci_lru(n - 2)

n = 100
start = time.time()
result = fibonacci_lru(n)
elapsed = time.time() - start

print(f"fibonacci({n}) = {result}")
print("time (seconds):", round(elapsed, 6))

print("cache info:", fibonacci_lru.cache_info())fibonacci_lru.cache_clear()
```

```python
# Memoization method 3: class-based wrapper
# A custom decorator class that caches results by storing (argument tuple → result) pairs.
# __call__ checks if arguments are cached; if not, computes and stores the result.
# This approach is more educational than practical but shows how decorators work under the hood.
# Useful for understanding decorator mechanics or when you need custom caching behavior.

class Memoize:
    def __init__(self, func):
        self.func = func
        self.cache = {}

    def __call__(self, *args):
        if args not in self.cache:
            self.cache[args] = self.func(*args)
        return self.cache[args]

@Memoize
def fibonacci_class(n):
    if n <= 1:
        return n
    return fibonacci_class(n - 1) + fibonacci_class(n - 2)

n = 100
start = time.time()
result = fibonacci_class(n)
elapsed = time.time() - start


print(f"fibonacci({n}) = {result}")print("cache size:", len(fibonacci_class.cache))
print("time (seconds):", round(elapsed, 6))
```

## Performance Summary

This table compares recursion and memoization approaches by approach, complexity, and practical usage:
- Naive Fibonacci: impractical for n > 35 due to exponential growth.
- Memoized approaches: practical for large n with dramatic speedup.
- Tail recursion: structurally efficient but Python doesn't optimize tail calls.
- Iterative DP: often fastest in Python since it avoids function call overhead entirely.

The goal is to highlight tradeoffs so you can pick the right approach for your constraints.
Use memoization or iterative methods when performance matters most.

| Approach | Time Complexity | Space Complexity | Notes |
|---|---|---|---|
| Fibonacci (naive) | O(2^n) | O(n) | Exponential, avoid for large n |
| Fibonacci (memoized) | O(n) | O(n) | Caches results |
| Tail recursion | O(n) | O(n) in Python | Python does not optimize tail calls |
| Iterative DP | O(n) | O(1) or O(n) | Often fastest in Python |
