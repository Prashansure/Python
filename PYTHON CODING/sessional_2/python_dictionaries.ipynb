## 1. Creating Dictionaries

Dictionaries store data as **key-value pairs**. Keys must be immutable (strings, numbers, tuples).

| Method | Syntax |
|--------|--------|
| Literal | `{'key': value}` |
| Constructor | `dict(key=value)` |
| From pairs | `dict([('a', 1), ('b', 2)])` |
| Comprehension | `{k: v for k, v in iterable}` |

```python
# Using curly braces (most common)
person = {'name': 'Alice', 'age': 30, 'city': 'Toronto'}
print(f"Using braces: {person}")

# Using dict() constructor
settings = dict(theme='dark', fontSize=14, autoSave=True)
print(f"Using dict(): {settings}")

# From list of tuples
pairs = [('a', 1), ('b', 2), ('c', 3)]
letters = dict(pairs)
print(f"From tuples: {letters}")

# Empty dictionary
empty = {}
also_empty = dict()
print(f"Empty dicts: {empty}, {also_empty}")

# Mixed key types (keys must be immutable)
mixed = {1: 'one', 'two': 2, (3, 4): 'tuple key'}
print(f"Mixed keys: {mixed}")
```

## 2. Accessing and Modifying Values

| Operation | Syntax | Notes |
|-----------|--------|-------|
| Get value | `d[key]` or `d.get(key)` | `[]` raises KeyError if missing |
| Set value | `d[key] = value` | Creates or updates |
| Delete | `del d[key]` | Raises KeyError if missing |
| Check key | `key in d` | Returns True/False |

```python
person = {'name': 'Alice', 'age': 30, 'city': 'Toronto'}

# Accessing values
print(f"Name: {person['name']}")
print(f"Age using get(): {person.get('age')}")

# get() with default value (safer - no KeyError)
print(f"Country: {person.get('country', 'Not specified')}")

# Modifying values
person['age'] = 31
print(f"Updated age: {person['age']}")

# Adding new key-value pairs
person['email'] = 'alice@email.com'
print(f"After adding email: {person}")

# Checking if key exists
print(f"'name' in person: {'name' in person}")
print(f"'phone' in person: {'phone' in person}")

# Deleting a key
del person['city']
print(f"After deleting city: {person}")
```

## 3. Dictionary Methods

| Method | Description |
|--------|-------------|
| `keys()` | Returns view of all keys |
| `values()` | Returns view of all values |
| `items()` | Returns view of (key, value) pairs |
| `get(key, default)` | Get value with optional default |
| `pop(key)` | Remove and return value |
| `popitem()` | Remove and return last item |
| `update(other)` | Merge another dict |
| `setdefault(key, value)` | Get or set if missing |
| `clear()` | Remove all items |
| `copy()` | Shallow copy |

```python
d = {'a': 1, 'b': 2, 'c': 3}

# keys(), values(), items()
print(f"Keys: {list(d.keys())}")
print(f"Values: {list(d.values())}")
print(f"Items: {list(d.items())}")

# pop() - remove and return
removed = d.pop('c')
print(f"Popped 'c': {removed}, dict now: {d}")

# update() - merge dictionaries
d.update({'d': 4, 'e': 5})
print(f"After update: {d}")

# setdefault() - get or set if missing
val = d.setdefault('f', 6)  # Sets 'f' to 6 since it doesn't exist
print(f"setdefault('f', 6): {val}, dict: {d}")
val = d.setdefault('a', 99)  # Returns 1, doesn't change since 'a' exists
print(f"setdefault('a', 99): {val}")

# copy() - shallow copy
d_copy = d.copy()
print(f"Copy: {d_copy}")
```

## 4. Iterating Over Dictionaries

Multiple ways to loop through dictionary contents.

```python
scores = {'Alice': 95, 'Bob': 87, 'Charlie': 92}

# Iterate over keys (default)
print("Keys:")
for name in scores:
    print(f"  {name}")

# Iterate over values
print("\nValues:")
for score in scores.values():
    print(f"  {score}")

# Iterate over key-value pairs (most common)
print("\nKey-Value pairs:")
for name, score in scores.items():
    print(f"  {name}: {score}")

# With enumerate for index
print("\nWith index:")
for i, (name, score) in enumerate(scores.items()):
    print(f"  {i}. {name} = {score}")
```

## 5. Dictionary Comprehensions

Concise way to create dictionaries, similar to list comprehensions.

**Syntax:** `{key_expr: value_expr for item in iterable if condition}`

```python
# Basic comprehension - squares
squares = {x: x**2 for x in range(1, 6)}
print(f"Squares: {squares}")

# With condition - even numbers only
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(f"Even squares: {even_squares}")

# From two lists using zip
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
name_to_age = {name: age for name, age in zip(names, ages)}
print(f"Name to age: {name_to_age}")

# Swap keys and values
original = {'a': 1, 'b': 2, 'c': 3}
swapped = {v: k for k, v in original.items()}
print(f"Swapped: {swapped}")

# Transform values
prices = {'apple': 1.50, 'banana': 0.75, 'orange': 2.00}
discounted = {item: price * 0.9 for item, price in prices.items()}
print(f"10% off: {discounted}")
```

## 6. Nested Dictionaries

Dictionaries can contain other dictionaries as values, useful for hierarchical data.

```python
# Nested dictionary structure
company = {
    'employees': {
        'E001': {'name': 'Alice', 'dept': 'Engineering'},
        'E002': {'name': 'Bob', 'dept': 'Marketing'}
    },
    'departments': ['Engineering', 'Marketing', 'HR']
}

# Accessing nested values
print(f"Employee E001: {company['employees']['E001']}")
print(f"E001's name: {company['employees']['E001']['name']}")

# Safe access with get()
dept = company.get('employees', {}).get('E003', {}).get('dept', 'Unknown')
print(f"E003's dept: {dept}")

# Modifying nested values
company['employees']['E001']['salary'] = 75000
print(f"After adding salary: {company['employees']['E001']}")

# Iterating nested dictionary
print("\nAll employees:")
for emp_id, emp_data in company['employees'].items():
    print(f"  {emp_id}: {emp_data['name']} - {emp_data['dept']}")
```

## 7. Common Dictionary Patterns

Useful patterns and techniques for working with dictionaries.

```python
# Pattern 1: Counting occurrences
words = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
counter = {}
for word in words:
    counter[word] = counter.get(word, 0) + 1
print(f"Word counts: {counter}")

# Pattern 2: Grouping by key
students = [('Alice', 'A'), ('Bob', 'B'), ('Charlie', 'A'), ('Diana', 'B')]
by_grade = {}
for name, grade in students:
    by_grade.setdefault(grade, []).append(name)
print(f"By grade: {by_grade}")

# Pattern 3: Merging dictionaries (Python 3.9+)
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
merged = dict1 | dict2  # Python 3.9+
print(f"Merged: {merged}")

# Pattern 4: Default values with collections.defaultdict
from collections import defaultdict
dd = defaultdict(list)
dd['fruits'].append('apple')
dd['fruits'].append('banana')
dd['vegetables'].append('carrot')
print(f"defaultdict: {dict(dd)}")
```

## 8. Practical Use Cases

Real-world applications of dictionaries.

```python
# Use Case 1: Configuration settings
config = {
    'debug': True,
    'database': {'host': 'localhost', 'port': 5432},
    'cache_timeout': 300
}
print(f"DB Host: {config['database']['host']}")

# Use Case 2: Caching/Memoization
cache = {}
def expensive_operation(n):
    if n in cache:
        return cache[n]
    result = n ** 2  # Simulating expensive calculation
    cache[n] = result
    return result

print(f"First call: {expensive_operation(10)}")
print(f"Cached call: {expensive_operation(10)}")

# Use Case 3: Lookup table / switch-case replacement
def get_day_type(day):
    day_types = {
        'Monday': 'Weekday', 'Tuesday': 'Weekday', 'Wednesday': 'Weekday',
        'Thursday': 'Weekday', 'Friday': 'Weekday',
        'Saturday': 'Weekend', 'Sunday': 'Weekend'
    }
    return day_types.get(day, 'Invalid day')

print(f"Saturday is: {get_day_type('Saturday')}")
```

## 9. Programming Exercises

Practice problems to reinforce dictionary concepts.

```python
# Problem 1: Word Frequency Counter
# Count occurrences of each word in a sentence
def word_frequency(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

# Test
print(word_frequency("the cat sat on the mat"))  # {'the': 2, 'cat': 1, ...}
```

```python
# Problem 2: Merge Dictionaries
# Merge two dictionaries, summing values for common keys
def merge_sum(dict1, dict2):
    result = dict1.copy()
    for key, value in dict2.items():
        result[key] = result.get(key, 0) + value
    return result

# Test
print(merge_sum({'a': 1, 'b': 2}, {'b': 3, 'c': 4}))  # {'a': 1, 'b': 5, 'c': 4}
```

```python
# Problem 3: Invert Dictionary
# Swap keys and values (assumes values are unique)
def invert_dict(d):
    return {v: k for k, v in d.items()}

# Test
print(invert_dict({'a': 1, 'b': 2, 'c': 3}))  # {1: 'a', 2: 'b', 3: 'c'}
```

```python
# Problem 4: Find Key with Max Value
# Return the key with the highest value
def key_with_max_value(d):
    return max(d, key=d.get)

# Test
print(key_with_max_value({'a': 10, 'b': 25, 'c': 15}))  # 'b'
```

```python
# Problem 5: Group by Length
# Group words by their length
def group_by_length(words):
    groups = {}
    for word in words:
        length = len(word)
        groups.setdefault(length, []).append(word)
    return groups

# Test
print(group_by_length(['cat', 'dog', 'elephant', 'ant', 'bear']))
# {3: ['cat', 'dog', 'ant'], 8: ['elephant'], 4: ['bear']}
```

```python
# Problem 6: Sort Dictionary by Value
# Return dictionary sorted by values
def sort_by_value(d, reverse=False):
    return dict(sorted(d.items(), key=lambda x: x[1], reverse=reverse))

# Test
print(sort_by_value({'a': 3, 'b': 1, 'c': 2}))  # {'b': 1, 'c': 2, 'a': 3}
print(sort_by_value({'a': 3, 'b': 1, 'c': 2}, reverse=True))  # {'a': 3, 'c': 2, 'b': 1}
```

```python
# Problem 7: Two Sum with Dictionary
# Find indices of two numbers that add up to target
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

# Test
print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
```

```python
# Problem 8: Character Frequency
# Count frequency of each character in a string
def char_frequency(s):
    freq = {}
    for char in s.lower():
        if char.isalpha():
            freq[char] = freq.get(char, 0) + 1
    return freq

# Test
print(char_frequency("Hello World"))  # {'h': 1, 'e': 1, 'l': 3, 'o': 2, ...}
```

```python
# Problem 9: Find Difference Between Dicts
# Return keys that differ between two dictionaries
def dict_difference(dict1, dict2):
    all_keys = set(dict1.keys()) | set(dict2.keys())
    diff = {}
    for key in all_keys:
        val1 = dict1.get(key)
        val2 = dict2.get(key)
        if val1 != val2:
            diff[key] = {'dict1': val1, 'dict2': val2}
    return diff

# Test
print(dict_difference({'a': 1, 'b': 2}, {'a': 1, 'b': 3, 'c': 4}))
# {'b': {'dict1': 2, 'dict2': 3}, 'c': {'dict1': None, 'dict2': 4}}
```

```python
# Problem 10: Flatten Nested Dictionary
# Convert nested dict to flat dict with dot notation keys
def flatten_dict(d, parent_key='', sep='.'):
    items = []
    for key, value in d.items():
        new_key = f"{parent_key}{sep}{key}" if parent_key else key
        if isinstance(value, dict):
            items.extend(flatten_dict(value, new_key, sep).items())
        else:
            items.append((new_key, value))
    return dict(items)

# Test
nested = {'a': 1, 'b': {'c': 2, 'd': {'e': 3}}}
print(flatten_dict(nested))  # {'a': 1, 'b.c': 2, 'b.d.e': 3}
```
