## 1. Creating Sets

Sets are unordered collections of unique elements. They are defined using curly braces `{}` or the `set()` constructor.

```python
# Different ways to create sets
empty_set = set()  # Note: {} creates an empty dict, not set!
numbers = {1, 2, 3, 4, 5}
mixed = {1, "hello", 3.14, True}

# From other iterables
from_list = set([1, 2, 2, 3, 3, 3])  # Duplicates removed
from_string = set("hello")  # Unique characters
from_tuple = set((1, 2, 3))

# Duplicates are automatically removed
with_duplicates = {1, 2, 2, 3, 3, 3, 4}

print(f"Empty set: {empty_set}")
print(f"Numbers: {numbers}")
print(f"Mixed types: {mixed}")
print(f"From list (duplicates removed): {from_list}")
print(f"From string: {from_string}")
print(f"With duplicates → {with_duplicates}")
```

## 2. Set Properties

Sets have three key properties:
- **Unordered**: Elements have no index/position
- **Unique**: No duplicate elements allowed
- **Mutable**: Can add/remove elements (but elements must be immutable)

```python
# Unordered - no indexing allowed
my_set = {3, 1, 4, 1, 5, 9, 2, 6}
print(f"Set: {my_set}")  # Order may vary
# print(my_set[0])  # TypeError: 'set' object is not subscriptable

# Unique - duplicates automatically removed
numbers = {1, 2, 2, 3, 3, 3}
print(f"Duplicates removed: {numbers}")

# Membership testing (very fast - O(1))
print(f"3 in set: {3 in my_set}")
print(f"10 in set: {10 in my_set}")

# Length
print(f"Length: {len(my_set)}")

# Elements must be hashable (immutable)
valid_set = {1, "hello", (1, 2, 3)}  # OK - all hashable
# invalid_set = {1, [2, 3]}  # TypeError: unhashable type: 'list'
```

## 3. Adding and Removing Elements

| Method | Description |
|--------|-------------|
| `add(x)` | Add element x |
| `remove(x)` | Remove x (raises KeyError if not found) |
| `discard(x)` | Remove x (no error if not found) |
| `pop()` | Remove and return arbitrary element |
| `clear()` | Remove all elements |

```python
# add() - add single element
fruits = {"apple", "banana"}
fruits.add("cherry")
print(f"After add: {fruits}")

# update() - add multiple elements
fruits.update(["date", "elderberry"])
print(f"After update: {fruits}")

# remove() vs discard()
fruits.remove("banana")  # Raises KeyError if not found
print(f"After remove: {fruits}")

fruits.discard("mango")  # No error if not found
print(f"After discard (non-existent): {fruits}")

# pop() - remove and return arbitrary element
popped = fruits.pop()
print(f"Popped: {popped}, Remaining: {fruits}")

# clear() - remove all elements
temp = {1, 2, 3}
temp.clear()
print(f"After clear: {temp}")
```

## 4. Set Operations

Sets support mathematical operations: union, intersection, difference, and symmetric difference.

| Operation | Method | Operator | Description |
|-----------|--------|----------|-------------|
| Union | `a.union(b)` | `a \| b` | All elements in a OR b |
| Intersection | `a.intersection(b)` | `a & b` | Elements in BOTH a AND b |
| Difference | `a.difference(b)` | `a - b` | Elements in a but NOT in b |
| Symmetric Diff | `a.symmetric_difference(b)` | `a ^ b` | Elements in a OR b but NOT both |

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

# Union - all elements from both sets
print(f"a | b (union): {a | b}")
print(f"a.union(b): {a.union(b)}")

# Intersection - common elements only
print(f"\na & b (intersection): {a & b}")
print(f"a.intersection(b): {a.intersection(b)}")

# Difference - in a but not in b
print(f"\na - b (difference): {a - b}")
print(f"b - a (difference): {b - a}")

# Symmetric difference - in a or b, but not both
print(f"\na ^ b (symmetric diff): {a ^ b}")
print(f"a.symmetric_difference(b): {a.symmetric_difference(b)}")
```

```python
# Visual representation of set operations
print("Set A: {1, 2, 3, 4, 5}")
print("Set B: {4, 5, 6, 7, 8}")
print()
print("A ∪ B (Union):        {1, 2, 3, 4, 5, 6, 7, 8}  - All elements")
print("A ∩ B (Intersection): {4, 5}                    - Common elements")
print("A - B (Difference):   {1, 2, 3}                 - Only in A")
print("B - A (Difference):   {6, 7, 8}                 - Only in B")
print("A △ B (Sym. Diff):    {1, 2, 3, 6, 7, 8}        - Not in both")
```

## 5. Set Methods Reference

| Method | Description |
|--------|-------------|
| `add(x)` | Add element x |
| `remove(x)` | Remove x (raises KeyError if missing) |
| `discard(x)` | Remove x (no error if missing) |
| `pop()` | Remove and return arbitrary element |
| `clear()` | Remove all elements |
| `copy()` | Return shallow copy |
| `update(other)` | Add elements from another iterable |
| `issubset(other)` | Check if set is subset |
| `issuperset(other)` | Check if set is superset |
| `isdisjoint(other)` | Check if no common elements |

```python
# Subset, superset, and disjoint checks
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}
c = {6, 7, 8}

print(f"a = {a}")
print(f"b = {b}")
print(f"c = {c}")
print()
print(f"a.issubset(b): {a.issubset(b)}")       # a ⊆ b
print(f"b.issuperset(a): {b.issuperset(a)}")   # b ⊇ a
print(f"a.isdisjoint(c): {a.isdisjoint(c)}")   # No common elements
print()

# Using operators for subset/superset
print(f"a <= b (subset): {a <= b}")
print(f"a < b (proper subset): {a < b}")
print(f"b >= a (superset): {b >= a}")
```

## 6. Set Comprehensions

Set comprehensions provide a concise way to create sets, similar to list comprehensions.

**Syntax:** `{expression for item in iterable if condition}`

```python
# Basic set comprehension
squares = {x**2 for x in range(1, 6)}
print(f"Squares: {squares}")

# With condition - even numbers only
evens = {x for x in range(10) if x % 2 == 0}
print(f"Even numbers: {evens}")

# From string - unique characters
text = "hello world"
unique_letters = {char for char in text if char.isalpha()}
print(f"Unique letters in '{text}': {unique_letters}")

# Converting list with duplicates
nums = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique_nums = {x for x in nums}
print(f"Unique from {nums}: {unique_nums}")
```

## 7. Frozen Sets

A **frozenset** is an immutable version of a set. Once created, elements cannot be added or removed.

**Use cases:**
- Dictionary keys (sets can't be used as keys)
- Elements of other sets
- When immutability is required

```python
# Creating a frozenset
fs = frozenset([1, 2, 3, 4, 5])
print(f"Frozenset: {fs}")
print(f"Type: {type(fs)}")

# Frozensets support set operations
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4, 5])
print(f"\nUnion: {fs1 | fs2}")
print(f"Intersection: {fs1 & fs2}")

# Using frozenset as dictionary key
permissions = {
    frozenset(['read']): 'Guest',
    frozenset(['read', 'write']): 'User',
    frozenset(['read', 'write', 'delete']): 'Admin'
}
print(f"\nPermissions dict: {permissions}")

# Set of frozensets
set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print(f"Set of frozensets: {set_of_sets}")
```

## 8. Practical Use Cases

Sets are ideal for:
- **Membership testing** - O(1) average lookup time
- **Removing duplicates** - Automatic deduplication
- **Mathematical operations** - Union, intersection, difference
- **Data validation** - Checking allowed values

```python
# Use Case 1: Fast membership testing
valid_statuses = {'active', 'pending', 'completed', 'cancelled'}
user_status = 'active'
print(f"Is '{user_status}' valid? {user_status in valid_statuses}")

# Use Case 2: Finding unique items
orders = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
unique_orders = set(orders)
print(f"Unique orders: {unique_orders}")

# Use Case 3: Finding common elements between lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common = set(list1) & set(list2)
print(f"Common elements: {common}")

# Use Case 4: Finding missing items
required = {'name', 'email', 'password'}
provided = {'name', 'email'}
missing = required - provided
print(f"Missing fields: {missing}")
```

## 9. Programming Exercises

Practice problems to reinforce set concepts.

```python
# Problem 1: Remove Duplicates
# Write a function that removes duplicates from a list using sets
def remove_duplicates(lst):
    return list(set(lst))

# Test
print(remove_duplicates([1, 2, 2, 3, 3, 3, 4]))  # [1, 2, 3, 4]
```

```python
# Problem 2: Find Common Elements
# Write a function that finds common elements between two lists
def find_common(list1, list2):
    return list(set(list1) & set(list2))

# Test
print(find_common([1, 2, 3, 4], [3, 4, 5, 6]))  # [3, 4]
```

```python
# Problem 3: Unique Characters
# Write a function that returns unique characters from a string
def unique_chars(s):
    return set(s.lower().replace(" ", ""))

# Test
print(unique_chars("Hello World"))  # {'h', 'e', 'l', 'o', 'w', 'r', 'd'}
```

```python
# Problem 4: Check Subset
# Write a function that checks if list1 is a subset of list2
def is_subset(list1, list2):
    return set(list1).issubset(set(list2))

# Test
print(is_subset([1, 2], [1, 2, 3, 4]))  # True
print(is_subset([1, 5], [1, 2, 3, 4]))  # False
```

```python
# Problem 5: Symmetric Difference
# Write a function that returns elements in either list but not both
def symmetric_diff(list1, list2):
    return list(set(list1) ^ set(list2))

# Test
print(symmetric_diff([1, 2, 3], [3, 4, 5]))  # [1, 2, 4, 5]
```

```python
# Problem 6: Count Unique Words
# Write a function that counts unique words in a sentence
def count_unique_words(sentence):
    words = sentence.lower().split()
    return len(set(words))

# Test
print(count_unique_words("the quick brown fox jumps over the lazy dog"))  # 8
```

```python
# Problem 7: Check Disjoint Lists
# Write a function that checks if two lists have no common elements
def are_disjoint(list1, list2):
    return set(list1).isdisjoint(set(list2))

# Test
print(are_disjoint([1, 2, 3], [4, 5, 6]))  # True
print(are_disjoint([1, 2, 3], [3, 4, 5]))  # False
```

```python
# Problem 8: Find Missing Elements
# Write a function that finds elements in list1 that are not in list2
def find_missing(list1, list2):
    return list(set(list1) - set(list2))

# Test
print(find_missing([1, 2, 3, 4, 5], [2, 4]))  # [1, 3, 5]
```

```python
# Problem 9: Union of Multiple Lists
# Write a function that returns union of multiple lists
def union_all(*lists):
    result = set()
    for lst in lists:
        result.update(lst)
    return list(result)

# Test
print(union_all([1, 2], [3, 4], [4, 5, 6]))  # [1, 2, 3, 4, 5, 6]
```

```python
# Problem 10: Find Duplicates in List
# Write a function that finds duplicate elements in a list
def find_duplicates(lst):
    seen = set()
    duplicates = set()
    for item in lst:
        if item in seen:
            duplicates.add(item)
        seen.add(item)
    return list(duplicates)

# Test
print(find_duplicates([1, 2, 3, 2, 4, 3, 5]))  # [2, 3]
```
